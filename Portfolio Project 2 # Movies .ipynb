```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("movies.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
print(df.isnull().sum())
```

```python
df.drop_duplicates()
```

```python
# for numerical values 

mean_value = df[['budget', 'gross']].mean()
df.loc[:, ['budget','gross']] = df.loc[:,['budget', 'gross']].fillna(value= mean_value)
print(df.isnull().sum())
```

```python
# EDA 

df.describe()
```

```python
sns.pairplot(df, kind='scatter', plot_kws = {'alpha': 0.5})
```

```python
df.corr(method = 'pearson', numeric_only = 'False')
```

```python
sns.heatmap(df.corr(method='pearson', numeric_only= 'False'))
```

```python
sns.lmplot(x='budget', y='gross', data=df, scatter_kws={'alpha': 0.3})
```

```python
sns.lmplot(x='votes',y='gross', data= df , scatter_kws ={'alpha': 0.3})
```
