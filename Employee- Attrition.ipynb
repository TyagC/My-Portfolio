```python
df_encoded = pd.get_dummies(df, drop_first=True)
df_encoded.info()
```

```python
correlation = df_encoded.drop('Attrition_Yes', axis=1).corrwith(df_encoded['Attrition_Yes'])
correlation.hvplot(kind= 'barh',figsize = (10,30))
```

```python
df_encoded.drop('Attrition_Yes', axis=1).corrwith(df_encoded.Attrition_Yes).sort_values().plot(kind='barh', figsize=(10, 30))
```

```python
col = df_encoded.corr().nlargest(20, "Attrition_Yes").Attrition_Yes.index
plt.figure(figsize=(15, 15))
sns.heatmap(df_encoded[col].corr(), annot=True, cmap="RdYlGn", annot_kws={"size":10})
```
